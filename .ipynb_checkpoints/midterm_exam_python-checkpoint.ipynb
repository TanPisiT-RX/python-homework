### ข้อ 2. จงเขียนโปรแกรมรับค่าชื่อและอายุจากผู้ใช้ แล้วแสดงผลตามรูปแบบที่กำหนด (2 คะแนน)

*ตัวอย่างการทำงาน:*
```
กรุณาป้อนชื่อ: สมชาย
กรุณาป้อนอายุ: 20

ผลลัพธ์:
ชื่อ: สมชาย
อายุ: 20 ปี
ปีเกิด พ.ศ.: 2547
```

````python
# พิมพ์โค้ดคำตอบที่นี่
name = input("กรุณาป้อนชื่อ")
age= int(input("กรุณาป้อนอายุ"))
current_year = 2569
brih_year= current_year-age

print("ผลลัพธ์:")
print("ชื่อ:", name)
print("อายุ",age,"ปี")
print("ปีเกิด พ.ศ.:",brih_year)
````

### ข้อ 3. จงเขียนโปรแกรมคำนวณพื้นที่สี่เหลี่ยมผืนผ้า โดยรับค่าความกว้างและความยาวจากผู้ใช้ (2 คะแนน)

*ตัวอย่างการทำงาน:*
```
ป้อนความกว้าง: 5
ป้อนความยาว: 8

ผลลัพธ์:
พื้นที่สี่เหลี่ยมผืนผ้า = 40.00 ตารางหน่วย
```

````python
# พิมพ์โค้ดคำตอบที่นี่
width =float(input("ป้อนความกว้าง"))
length = float(input("ป้อนความยาว: "))
area = width * length

print("ผลลัพธ์:")
print(f"พื้นที่สี่เหี่ยมผืนผ้า={area:.2f}ตารางหน่วย")
````

### ข้อ 4. กำหนดให้ List ของคะแนนสอบนักศึกษา ดังนี้ (2 คะแนน)

```python
scores = [78, 85, 92, 64, 55, 88, 71, 95, 42, 80]
```

จงเขียนโปรแกรมเพื่อ:
- a) แสดงคะแนนเฉลี่ย (ทศนิยม 2 ตำแหน่ง)

````python
scores = [78, 85, 92, 64, 55, 88, 71, 95, 42, 80]

# a) แสดงคะแนนเฉลี่ย (ทศนิยม 2 ตำแหน่ง)
average = sum(scores) / len(scores)

print(f"คะแนนเฉลี่ย={average:.2f}")
````

### ข้อ 5. กำหนดให้ List ของชื่อนักศึกษา ดังนี้ (2 คะแนน)

```python
students = ["สมชาย", "สมหญิง", "สมศรี", "สมศักดิ์", "สมปอง"]
```

จงเขียนโปรแกรมเพื่อ:
- a) เพิ่มชื่อ "สมพร" ต่อท้าย List
- b) แทรกชื่อ "สมใจ" ในตำแหน่งที่ 2 (index = 2)
- c) ลบชื่อ "สมศรี" ออกจาก List
- d) แสดง List ที่อัปเดตแล้ว และแสดงจำนวนสมาชิกทั้งหมด

````python
students = ["สมชาย", "สมหญิง", "สมศรี", "สมศักดิ์", "สมปอง"]

# a) เพิ่มชื่อ "สมพร" ต่อท้าย List
students.append("สมพร")

# b) แทรกชื่อ "สมใจ" ในตำแหน่งที่ 2 (index = 2)
students.insert(2, "สมใจ")

# c) ลบชื่อ "สมศรี" ออกจาก List
students.remove("สมศรี")

# d) แสดง List ที่อัปเดตแล้ว และแสดงจำนวนสมาชิกทั้งหมด

print(students)
print("จำนวนสมาชิกทั้งหมด:", len(students))
````

### ข้อ 6. กำหนดให้ `msg = "Python Programming"` จงเขียนโค้ดเพื่อแสดงผลลัพธ์ดังนี้ (2 คะแนน)

### ข้อ 7. จงแปลงข้อมูลต่อไปนี้ให้อยู่ในรูปแบบ Dictionary และแสดงวิธีเข้าถึงข้อมูลของ "นายสมชาย" (2 คะแนน)

| รหัส | ชื่อ | อายุ | เกรดเฉลี่ย |
|:---:|:---:|:---:|:---:|
| 6601 | นายสมชาย | 20 | 3.25 |
| 6602 | นางสาวสมหญิง | 19 | 3.50 |

````python
# สร้าง Dictionary
students = {
    "6601": {"ชื่อ": "นายสมชาย", "อายุ": 20, "เกรดเฉลี่ย": 3.25},
    "6602": {"ชื่อ": "นางสาวสมหญิง", "อายุ": 19, "เกรดเฉลี่ย": 3.50}
}
# แสดงวิธีเข้าถึงข้อมูลของ "นายสมชาย"
print("ชื่อ:", students["6601"]["ชื่อ"])
print("อายุ:", students["6601"]["อายุ"])
print("เกรดเฉลี่ย:", students["6601"]["เกรดเฉลี่ย"])
````

### ข้อ 8. จงเขียนโปรแกรมคำนวณพื้นที่วงกลมจากสูตร A = πr² โดยกำหนดให้ r = 7 และใช้ค่า π จาก math module (2 คะแนน)

````python
# พิมพ์โค้ดคำตอบที่นี่
import math
r = 7
area = math.pi * (r ** 2)

print(f"รัศมีวงกลม r = {r}")
print(f"พื้นที่วงกลม A = {area:.2f}")
````

### ข้อ 14. จงเขียนโปรแกรม FizzBuzz โดยแสดงเลข 1-30 ตามเงื่อนไขดังนี้ (2 คะแนน)

- ถ้าหารด้วย 3 ลงตัว ให้พิมพ์ "Fizz"
- ถ้าหารด้วย 5 ลงตัว ให้พิมพ์ "Buzz"
- ถ้าหารด้วย 3 และ 5 ลงตัว ให้พิมพ์ "FizzBuzz"
- กรณีอื่นๆ ให้พิมพ์ตัวเลขปกติ

*ตัวอย่างผลลัพธ์: 1, 2, Fizz, 4, Buzz, Fizz, 7, 8, Fizz, Buzz, 11, Fizz, 13, 14, FizzBuzz, ...*

````python
# พิมพ์โค้ดคำตอบที่นี่
for i in range(1, 31):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
````

### ข้อ 15. จงเขียนโปรแกรมแสดงรูปแบบดังต่อไปนี้ โดยรับค่า n จากผู้ใช้ (2 คะแนน)
