```python
class Dog():
    def __init__(self,name,gender,gene,age,weight,color):
        self.name = name
        self.gender = gender
        self.gene = gene
        self.age = age
        self.weight = weight
        self.color = color
    def sit(self):
        pass
    def sleep(self):
        pass
    def eat(self):
        pass
    def showinfo(self):
        print("""
        ชื่อ: {}
        เพศ: {}
        พันธุ์: {}
        อายุ: {}
        นํ้าหนัก: {}
        สี: {}
        """.format(self.name,self.gender,self.gene,self.age,self.weight,self.color))
```

```python
wed = Dog("bo","Male","thai ridgeback",7,50,"brown ")
wed.showinfo()
```

```python
class Student:
    def __init__(self, student_id, national_id, first_name, last_name, age, gender):
        self.student_id = student_id          
        self.national_id = national_id        
        self.first_name = first_name         
        self.last_name = last_name            
        self.age = age                       
        self.gender = gender 
        
    def show_info(self):
        print(f"รหัสนักศึกษา: {self.student_id}")
        print(f"เลขบัตรประชาชน: {self.national_id}")
        print(f"ชื่อ-นามสกุล: {self.first_name} {self.last_name}")
        print(f"อายุ: {self.age}")
        print(f"เพศ: {self.gender}")
        
```

```python
ggwp = Student("6812732110", "1211212121212", "Tanpisit", "Phomlok", 21, "male")
ggwp.show_info()
```
